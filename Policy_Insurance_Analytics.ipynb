```python

# Converting date columns into datetime format

policy['Policy Start Date'] = pd.to_datetime(
    policy['Policy Start Date'], errors='coerce'
)

policy['Policy End Date'] = pd.to_datetime(
    policy['Policy End Date'], errors='coerce'
)

claims['Date of Claim'] = pd.to_datetime(
    claims['Date of Claim'], errors='coerce'
)

claims['Settlement Date'] = pd.to_datetime(
    claims['Settlement Date'], errors='coerce'
)

payment['Date of Payment'] = pd.to_datetime(
    payment['Date of Payment'], errors='coerce'
)
```

```python

print("POLICY DATES")
print(policy[['Policy Start Date', 'Policy End Date']].dtypes)

print("\nCLAIM DATES")
print(claims[['Date of Claim', 'Settlement Date']].dtypes)

print("\nPAYMENT DATES")
print(payment['Date of Payment'].dtype)
```

```python

# Identifying missing settlement dates

claims['Settlement Date Missing'] = claims['Settlement Date'].isnull()

# Checking the result

print(claims['Settlement Date Missing'].value_counts())
```

```python

# Percentage of missing settlement dates by claim status

settlement_analysis = pd.crosstab(
    claims['Claim Status'],
    claims['Settlement Date Missing'],
    normalize='index'
) * 100

print(settlement_analysis.round(2))
```

```python

# Text cleaning

datasets = [customer, policy, claims, payment, additional]

for df in datasets:
    for col in df.select_dtypes(include='object').columns:
        df[col] = df[col].str.strip()

print("Text cleaning completed successfully!")
```

```python

# KPI Calculations

total_customers = customer['Customer ID'].nunique()

total_policies = policy['Policy ID'].nunique()

active_policies = (policy['Status'] == 'Active').sum()

total_premium = policy['Premium Amount'].sum()

total_claims = claims['Claim ID'].nunique()

approved_claims = (claims['Claim Status'] == 'Approved').sum()

total_payments = payment['Amount Paid'].sum()

# Display KPI values

print("POLICY INSURANCE KPI SUMMARY")
print("--------------------------------")
print("Total Customers:", total_customers)
print("Total Policies:", total_policies)
print("Active Policies:", active_policies)
print("Total Premium Amount:", round(total_premium, 2))
print("Total Claims:", total_claims)
print("Approved Claims:", approved_claims)
print("Total Payments:", round(total_payments, 2))
```

```python

import matplotlib.pyplot as plt

# KPI details
kpis = [
    ("Total Customers", total_customers),
    ("Total Policies", total_policies),
    ("Active Policies", active_policies),
    ("Total Premium", f"{total_premium:,.0f}"),
    ("Total Claims", total_claims),
    ("Approved Claims", approved_claims)
]

fig, axes = plt.subplots(2, 3, figsize=(15, 7))

fig.suptitle(
    "POLICY INSURANCE ANALYTICS DASHBOARD",
    fontsize=18,
    fontweight='bold'
)

for ax, (title, value) in zip(axes.flatten(), kpis):

    ax.set_facecolor('#F2EDE5')

    ax.text(
        0.5, 0.65, title,
        ha='center',
        fontsize=13,
        fontweight='bold',
        color='#75543D'
    )

    ax.text(
        0.5, 0.35, str(value),
        ha='center',
        fontsize=23,
        fontweight='bold',
        color='#3D3028'
    )

    ax.set_xticks([])
    ax.set_yticks([])

    for spine in ax.spines.values():
        spine.set_color('#B89B7A')
        spine.set_linewidth(2)

plt.tight_layout()
plt.show()
```

```python

import matplotlib.pyplot as plt

policy_counts = policy['Policy Type'].value_counts()

plt.figure(figsize=(9, 5))

plt.bar(
    policy_counts.index,
    policy_counts.values,
    color=['#806044', '#B89B7A', '#D8C3A5', '#A67C52']
)

plt.title('Policy Type Distribution',
          fontsize=16, fontweight='bold')

plt.xlabel('Policy Type')
plt.ylabel('Number of Policies')

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Claim Status Analysis

claim_counts = claims['Claim Status'].value_counts()

plt.figure(figsize=(9, 5))

plt.bar(
    claim_counts.index,
    claim_counts.values,
    color=['#806044', '#B89B7A', '#D8C3A5']
)

plt.title('Claim Status Distribution',
          fontsize=16, fontweight='bold')

plt.xlabel('Claim Status')
plt.ylabel('Number of Claims')

# Display values above bars
for i, value in enumerate(claim_counts.values):
    plt.text(i, value + 10, str(value),
             ha='center', fontweight='bold')

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Premium Amount by Policy Type

premium_analysis = policy.groupby('Policy Type')['Premium Amount'].sum()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    premium_analysis.index,
    premium_analysis.values,
    color=['#806044', '#B89B7A', '#D8C3A5', '#A67C52']
)

plt.title('Total Premium Amount by Policy Type',
          fontsize=16, fontweight='bold')

plt.xlabel('Policy Type')
plt.ylabel('Total Premium Amount')

# Display values
for bar in bars:
    plt.text(
        bar.get_x() + bar.get_width()/2,
        bar.get_height(),
        f'{bar.get_height():,.0f}',
        ha='center',
        va='bottom',
        fontsize=9
    )

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Monthly Claim Trend

claims['Date of Claim'] = pd.to_datetime(claims['Date of Claim'])

monthly_claims = (
    claims.groupby(claims['Date of Claim'].dt.to_period('M'))
    .size()
)

plt.figure(figsize=(12, 5))

plt.plot(
    monthly_claims.index.astype(str),
    monthly_claims.values,
    marker='o',
    color='#806044',
    linewidth=2
)

plt.title('Monthly Claim Trend', fontsize=16, fontweight='bold')
plt.xlabel('Month')
plt.ylabel('Number of Claims')
plt.xticks(rotation=45)
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Claim Status by Gender

gender_claims = pd.crosstab(
    customer['Gender'],
    claims['Claim Status']
)

gender_claims.plot(
    kind='bar',
    figsize=(10, 5),
    color=['#806044', '#C1A080', '#DDC9AC']
)

plt.title('Claim Status by Gender', fontsize=16, fontweight='bold')
plt.xlabel('Gender')
plt.ylabel('Number of Claims')
plt.xticks(rotation=0)
plt.legend(title='Claim Status')
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Payment Method Distribution

payment_counts = payment['Payment Method'].value_counts()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    payment_counts.index,
    payment_counts.values,
    color=['#806044', '#C1A080', '#DDC9AC']
)

plt.title('Payment Method Distribution',
          fontsize=16, fontweight='bold')

plt.xlabel('Payment Method')
plt.ylabel('Number of Transactions')

plt.bar_label(bars, padding=3)

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Settlement Date Analysis by Claim Status

settlement_analysis.plot(
    kind='bar',
    figsize=(9, 5),
    color=['#806044', '#C1A080']
)

plt.title('Settlement Date Analysis by Claim Status',
          fontsize=15, fontweight='bold')

plt.xlabel('Claim Status')
plt.ylabel('Percentage of Claims')

plt.xticks(rotation=0)
plt.legend(['Settlement Date Available', 'Settlement Date Missing'])

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Claim Amount by Policy Type

claim_policy = pd.merge(
    claims,
    policy[['Policy ID', 'Policy Type']],
    on='Policy ID',
    how='left'
)

claim_amount = claim_policy.groupby('Policy Type')['Claim Amount'].sum()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    claim_amount.index,
    claim_amount.values,
    color=['#806044', '#C1A080', '#DDC9AC', '#A77D56']
)

plt.title('Total Claim Amount by Policy Type',
          fontsize=16, fontweight='bold')

plt.xlabel('Policy Type')
plt.ylabel('Total Claim Amount')

plt.bar_label(bars, fmt='{:,.0f}', padding=3)

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Claim Status by Policy Type

claim_status_policy = pd.crosstab(
    claim_policy['Policy Type'],
    claim_policy['Claim Status']
)

claim_status_policy.plot(
    kind='bar',
    figsize=(10, 5),
    color=['#806044', '#C1A080', '#DDC9AC']
)

plt.title('Claim Status by Policy Type',
          fontsize=16, fontweight='bold')

plt.xlabel('Policy Type')
plt.ylabel('Number of Claims')

plt.xticks(rotation=0)
plt.legend(title='Claim Status')
plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python

# Average Claim Amount by Claim Status

avg_claim = claims.groupby('Claim Status')['Claim Amount'].mean()

plt.figure(figsize=(9, 5))

bars = plt.bar(
    avg_claim.index,
    avg_claim.values,
    color=['#806044', '#C1A080', '#DDC9AC']
)

plt.title('Average Claim Amount by Claim Status',
          fontsize=16, fontweight='bold')

plt.xlabel('Claim Status')
plt.ylabel('Average Claim Amount')

plt.bar_label(bars, fmt='{:,.0f}', padding=3)

plt.grid(axis='y', alpha=0.2)
plt.tight_layout()
plt.show()
```

```python
# KPI 5: Average Coverage Amount per Policy

avg_coverage = policy['Coverage Amount'].mean()

print("Average Coverage Amount per Policy:",
      round(avg_coverage, 2))
```

```python
# KPI 6: Average Premium Amount per Policy

avg_premium = policy['Premium Amount'].mean()

print("Average Premium Amount per Policy:",
      round(avg_premium, 2))
```

```python
# KPI 7: Active Policy Percentage

active_policies = (policy['Status'] == 'Active').sum()
total_policies = len(policy)

active_percentage = (active_policies / total_policies) * 100

print("Active Policy Percentage:",
      round(active_percentage, 2), "%")
```

```python
# KPI 8: Number of Policies by Status

policy_status_count = policy['Status'].value_counts()

print(policy_status_count)
```

```python
import matplotlib.pyplot as plt

active = (policy['Status'] == 'Active').sum()
inactive = (policy['Status'] != 'Active').sum()

print("Active Policies:", active)
print("Inactive Policies:", inactive)
print("Active : Inactive Ratio:", round(active / inactive, 2), ": 1")

plt.figure(figsize=(7, 5))

plt.bar(
    ['Active', 'Inactive'],
    [active, inactive],
    color=['#8B5E3C', '#D8C3A5'],
    edgecolor='#5C4033'
)

plt.title('Active vs Inactive Policies', fontweight='bold')
plt.xlabel('Policy Status')
plt.ylabel('Number of Policies')

for i, value in enumerate([active, inactive]):
    plt.text(i, value + 40, str(value), ha='center', fontweight='bold')

plt.tight_layout()
plt.show()
```
